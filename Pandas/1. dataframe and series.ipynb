### Series and Dataframe

```python
import pandas as pd

# Series (1D)
# Series is like a column in a table, all of elements in a Series must be of the same type.
s = pd.Series([10, 20, 30], index=["a", "b", "c"])

# DataFrame (2D)
# DataFrame is like a table, it can contain multiple Series (columns) with different types.
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Cara", "Dan"],
    "age": [25, 31, 29, 40],
    "city": ["Hanoi", "HCMC", "Da Nang", "Hanoi"],
    "income": [15_000_000, 22_000_000, None, 18_500_000],
})

# Display the first few rows of the DataFrame
df.head()
```

```python
print(type(df['name']))
# A Series is a single column in a DataFrame, so df['name'] returns a Series.

print(type(df[['name', 'age']]))
# A DataFrame is a table with multiple columns, so df[['name', 'age']] returns a DataFrame with just those two columns.
```

### How to create a dataframe

```python
# Help me to give me some ways to create a DataFrame
# 1. From a dictionary of lists
data = {
    "name": ["Alice", "Bob", "Cara"],
    "age": [25, 31, 29],
    "city": ["Hanoi", "HCMC", "Da Nang"]
}
df_from_dict = pd.DataFrame(data)

# 2. From a list of dictionaries
data_list = [
    {"name": "Alice", "age": 25, "city": "Hanoi"},
    {"name": "Bob", "age": 31, "city": "HCMC"},
    {"name": "Cara", "age": 29, "city": "Da Nang"}
]
df_from_list = pd.DataFrame(data_list)

# 3. From a NumPy array
import numpy as np
data_array = np.array([
    ["Alice", 25, "Hanoi"],
    ["Bob", 31, "HCMC"],
    ["Cara", 29, "Da Nang"]
])
df_from_array = pd.DataFrame(data_array, columns=["name", "age", "city"])   

# 4. From a CSV file
df_from_csv = pd.read_csv('path_to_file.csv')

# 5. From an Excel file
df_from_excel = pd.read_excel('path_to_file.xlsx')   

# 6. From a SQL database
import sqlite3
conn = sqlite3.connect('database.db')
df_from_sql = pd.read_sql_query("SELECT * FROM table_name", conn)
conn.close()

# 7. From a JSON file
df_from_json = pd.read_json('path_to_file.json')

# 9. From a dictionary of Series
data_series = {
    "name": pd.Series(["Alice", "Bob", "Cara"]),
    "age": pd.Series([25, 31, 29]),
    "city": pd.Series(["Hanoi", "HCMC", "Da Nang"])
}
df_from_series_dict = pd.DataFrame(data_series)

# 10. From a list of tuples
data_tuples = [
    ("Alice", 25, "Hanoi"),
    ("Bob", 31, "HCMC"),
    ("Cara", 29, "Da Nang")
]
df_from_tuples = pd.DataFrame(data_tuples, columns=["name", "age", "city"])

# 11. From a dictionary of NumPy arrays
data_numpy = {
    "name": np.array(["Alice", "Bob", "Cara"]),
    "age": np.array([25, 31, 29]),
    "city": np.array(["Hanoi", "HCMC", "Da Nang"])
}
df_from_numpy = pd.DataFrame(data_numpy)

# 12. From a dictionary of lists with different lengths
data_diff_length = {
    "name": ["Alice", "Bob", "Cara"],
    "age": [25, 31],
    "city": ["Hanoi", "HCMC", "Da Nang"]
}
# This will fill missing values with NaN
df_from_diff_length = pd.DataFrame(data_diff_length)
```
